     ABSTRACTION:

It means hiding complex implementation details and showing only the essential features of the object.

In Python, abstraction is achieved using abstract classes and abstract methods, which are defined using the abc module (Abstract Base Class).

     🔹When to Use Abstraction?

When you want to define a blueprint or structure for other classes.

When some methods must be implemented in child classes, but not in the base class.

     🔹Summary

Concept	             Description

1.ABC	-              Abstract Base Class

2.@abstractmethod	-    Marks a method as abstract (must be overridden)

3.Can't instantiate	 -   You cannot create an object of an abstract directly

4.Must Override	-    Child classes must override all abstract methods


```python
from abc import ABC, abstractmethod 

# Abstract class
class Animal(ABC):
    
    @abstractmethod  # Proper decorator
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog says Woof!")

d = Dog()
d.sound()

```

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):
    
    
    def withdraw(self):
        pass

class SavingAccount(BankAccount):
    def withdraw(self):
        print("Withdrawn from Saving Account")

class CurrentAccount(BankAccount):
    def withdraw(self):
        print("Withdrawn from Current Account")

s = SavingAccount()
s.withdraw()

c = CurrentAccount()
c.withdraw()
```

```python
from abc import ABC, abstractmethod

class Remote(ABC):

    
    def power_button(self):
        pass

class TV(Remote):
    def power_button(self):
        print("TV turned ON/OFF")

class AC(Remote):
    def power_button(self):
        print("AC turned ON/OFF")

tv = TV()
tv.power_button()

ac = AC()
ac.power_button()
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Shape(ABC):
    
    @abstractmethod 

    def area(self):
        pass

# Concrete class
class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

# Using the class
s = Square(4)
print("Area of square:", s.area())
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

# Concrete classes
class Car(Vehicle):
    def start_engine(self):
        print("Car engine started.")

class Bike(Vehicle):
    def start_engine(self):
        print("Bike engine started.")

# Usage
car = Car()
car.start_engine()

bike = Bike()
bike.start_engine()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def make_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def make_payment(self, amount):
        print(f"Payment of ₹{amount} made using Credit Card.")

class UpiPayment(Payment):
    def make_payment(self, amount):
        print(f"Payment of ₹{amount} made using UPI.")

# Usage
p1 = CreditCardPayment()
p1.make_payment(500)

p2 = UpiPayment()
p2.make_payment(300)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Lion(Animal):
    def sound(self):
        print("Lion roars")

class Snake(Animal):
    def sound(self):
        print("Snake hisses")

class Cow(Animal):
    def sound(self):
        print("Cow says Moo")

# Usage
lion = Lion()
lion.sound()

snake = Snake()
snake.sound()

cow = Cow()
cow.sound()
```

```python
from abc import ABC, abstractmethod

class Student(ABC):
    @abstractmethod
    def get_result(self):
        pass

class ScienceStudent(Student):
    def get_result(self):
        print("Science student passed with 85%")

class ArtsStudent(Student):
    def get_result(self):
        print("Arts student passed with 78%")

# Usage
sci = ScienceStudent()
sci.get_result()

arts = ArtsStudent()
arts.get_result()
```
