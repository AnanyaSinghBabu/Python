MULTI-LEVEL INHERITANCE:

```python
# Base class (Grandparent)
class Animal:
    def eat(self):
        print("Animal eats")

# Derived class (Parent)
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Derived class (Child of Dog)
class Puppy(Dog):
    def weep(self):
        print("Puppy weeps")

# Create object of Puppy
p = Puppy()
p.eat()   # Inherited from Animal
p.bark()  # Inherited from Dog
p.weep()  # Defined in Puppy
```

```python
# Base class
class Person:
    def show_name(self):
        print("I am a person")

# Intermediate class
class Employee(Person):
    def show_job(self):
        print("I am an employee")

# Final derived class
class Manager(Employee):
    def show_role(self):
        print("I am a manager")

m = Manager()
m.show_name()  # From Person
m.show_job()   # From Employee
m.show_role()  # From Manager

```

MULTIPLE INHERITANCE:

In multiple inheritance, a class inherits from more than one parent class.


```python
# First parent class
class Father:
    def skills(self):
        print("Father: Gardening, Cooking")

# Second parent class
class Mother:
    def skills(self):
        print("Mother: Singing, Dancing")

# Child class inherits from both Father and Mother
class Child(Father, Mother):
    def skills(self):
        print("Child: Painting")
        Father.skills(self)   # calling Father’s skills
        Mother.skills(self)   # calling Mother’s skills

# Create object of Child
c = Child()
c.skills()
```

```python
# First parent class
class Student:
    def study(self):
        print("Student studies subjects.")

# Second parent class
class Athlete:
    def train(self):
        print("Athlete trains every day.")

# Child class inherits both
class Scholar(Student, Athlete):
    def activities(self):
        print("Scholar does both study and training.")

s = Scholar()
s.activities()
s.study()   # from Student
s.train()   # from Athlete
```

HIERARCHICAL INHERITANCE:

In hierarchical inheritance, multiple child classes inherit from a single parent class.

```python
# Base class
class Animal:
    def sound(self):
        print("Animals make sounds")

# Derived class 1
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Derived class 2
class Cat(Animal):
    def meow(self):
        print("Cat meows")

# Create object of Dog
d = Dog()
d.sound()  # from Animal
d.bark()   # from Dog

# Create object of Cat
c = Cat()
c.sound()  # from Animal
c.meow()   # from Cat
```

```python
# Base class
class Person:
    def show(self):
        print("I am a person")

# Derived class 1
class Teacher(Person):
    def teach(self):
        print("I can teach")

# Derived class 2
class Doctor(Person):
    def treat(self):
        print("I can treat patients")

# Create objects
t = Teacher()
t.show()
t.teach()

d = Doctor()
d.show()
d.treat()
```

HYBRID INHERITANCE:

 (Combination of Multilevel + Multiple)

```python
# Base class
class Person:
    def show_person(self):
        print("I am a Person")

# First derived class (Single inheritance)
class Student(Person):
    def show_student(self):
        print("I am a Student")

# Another base class
class Sports:
    def show_sports(self):
        print("I play Sports")

# Final class (inherits from Student and Sports) => Multiple + Multilevel
class SchoolCaptain(Student, Sports):
    def show_captain(self):
        print("I am the School Captain")

# Create object
captain = SchoolCaptain()
captain.show_person()   # From Person
captain.show_student()  # From Student
captain.show_sports()   # From Sports
captain.show_captain()  # Own method
```

```python
# Base class
class Person:
    def get_person(self):
        print("Person details")

# First child of Person
class Engineer(Person):
    def get_engineer(self):
        print("Engineer details")

# Second child of Person
class Artist(Person):
    def get_artist(self):
        print("Artist details")

# Hybrid class (inherits from both Engineer and Artist)
class CreativeEngineer(Engineer, Artist):
    def get_creative(self):
        print("Creative Engineer details")

ce = CreativeEngineer()
ce.get_person()     # From Person
ce.get_engineer()   # From Engineer
ce.get_artist()     # From Artist
ce.get_creative()   # Own method
```

Method Overriding & Constructor Overriding

They both occur in inheritance, especially when:

A child class redefines a method or constructor of the parent class.

Most commonly used in single, multilevel, and hierarchical inheritance.

Can also be used in multiple and hybrid inheritance, but needs careful handling.

1. Method Overriding → Same method name in parent and child class

🔷 Happens in:

✅ Single, Multilevel, Hierarchical, Hybrid, and Multiple inheritance

```python
class Parent:
    def show(self):
        print("Parent show method")

class Child(Parent):
    def show(self):  # Overriding the parent method
        print("Child show method")

c = Child()
c.show()  # Output: Child show method
```

```python
#  Method Overriding in Multilevel Inheritance

class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")  # overrides Animal's speak()

class Puppy(Dog):
    def speak(self):
        print("Puppy yelps")  # overrides Dog's speak()

p = Puppy()
p.speak()
```

 2. Constructor Overriding → Redefining __init__() in child class

🔷 Happens in:

✅ Single, Multilevel, Hierarchical, Hybrid, and Multiple inheritance

```python
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        print("Child constructor")  # Overrides parent constructor

c = Child()  # Output: Child constructor
```

```python
# Method Overriding in Multiple Inheritance

class A:
    def show(self):
        print("Show from A")

class B:
    def show(self):
        print("Show from B")

class C(A, B):
    def show(self):
        print("Show from C")
        A.show(self)
        B.show(self)

c = C()
c.show()
```

```python
# Constructor Overriding in Multiple Inheritance
class Father:
    def __init__(self):
        print("Father's constructor")

class Mother:
    def __init__(self):
        print("Mother's constructor")

class Child(Father, Mother):
    def __init__(self):
        Father.__init__(self)
        Mother.__init__(self)
        print("Child's constructor")

c = Child()
```
