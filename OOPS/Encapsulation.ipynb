Encapsulation is one of the core concepts of Object-Oriented Programming (OOP) in Python. It refers to bundling data (variables) and methods (functions) together within a class and restricting direct access to some of the object's components. This helps protect the data from unintended modifications and improves code maintainability.

      Why Use Encapsulation?
Data Protection: Prevent accidental changes to internal variables.

Control Access: Control who can read or modify attributes.

Cleaner Code: Hide unnecessary implementation details from the user.

     Python Implementation
Python uses naming conventions to indicate the visibility of variables/methods:

| Type      | Syntax Example | Accessibility                            |
| --------- | -------------- | ---------------------------------------- |
| Public    | `self.name`    | Can be accessed anywhere                 |
| Protected | `self._name`   | Internal use; suggest no outside access  |
| Private   | `self.__name`  | Name mangled; can't be accessed directly |


```python
class Public:
    def __init__(self):
        self.name = "John"  # Public attribute

    def display_name(self):
        print(self.name)  # Public method

obj = Public()
obj.display_name()  # Accessible
print(obj.name)     # Accessible
```

```python
class Protected:
    def __init__(self):
        self._age = 30  # Protected attribute

class Subclass(Protected):
    def display_age(self):
        print(self._age)  # Accessible in subclass

obj = Subclass()
obj.display_age()
```

```python
class Private:
    def __init__(self):
        self.__salary = 50000  # Private attribute

    def salary(self):
        return self.__salary  # Access through public method

obj = Private()
print(obj.salary())  # Works
#print(obj.__salary)  # Raises AttributeError
```

     Getter and Setter Methods
In Python, getter and setter methods are used to access and modify private attributes safely. Instead of accessing private data directly, these methods provide controlled access, allowing you to:

1.Read data using a getter method.

2.Update data using a setter method with optional validation or restrictions.

```python
class Employee:
    def __init__(self):
        self.__salary = 50000  # Private attribute

    def get_salary(self):    # Getter method
        return self.__salary

    def set_salary(self, amount):   # Setter method
        if amount > 0:
            self.__salary = amount
        else:
            print("Invalid salary amount!")

emp = Employee()
print(emp.get_salary())  # Access salary using getter

emp.set_salary(60000)   # Update salary using setter
print(emp.get_salary())
```
