    In Seaborn,a scatter plot is used to show the relationship between two continuous variables by displaying each observation as a point on an X–Y plane.

    Places each data point at the intersection of:

X-axis value (first variable)

Y-axis value (second variable)

     Optionally uses:

hue → to color points by category

style → to change marker shape by category

size → to make point size represent another variable

markers=True → Add markers at data points

```python
# 1. Scatter Plot - bill length vs bill depth
sns.scatterplot (
    x="bill_length_mm",           # Column name for X-axis values
    y="bill_depth_mm",            # Column name for Y-axis values
    hue="species",                # Color points based on penguin species
    style="species",              # Shape by species
    data=penguins )               # Dataset to use (a DataFrame)

plt.show()
```

    In Seaborn, a line plot is used to visualize the relationship between two continuous variables, where one (usually on the X-axis) represents some sort of ordered sequence — such as time, measurements, or an index — and the other (Y-axis) represents the value you’re measuring.

    What it does
    
Draws a line connecting data points in order of the X-axis variable.

    Often used to:

Show trends over time

Compare changes between categories

Highlight patterns in continuous data

```python
# 2. Line Plot - example with flipper length vs body mass
sns.lineplot(
x="flipper_length_mm", 
y="body_mass_g", 
hue="species",
data=penguins)
plt.show()
```

    In Seaborn, a histogram is used to show the distribution of a single variable by grouping the data into intervals (called bins) and counting how many data points fall into each interval.

    Divides the range of your data into equal-width bins (intervals).

Counts the number of observations in each bin.

    Draws bars where:

The x-axis = the bin ranges (value intervals).

The y-axis = count (or proportion) of observations in each bin.

Optionally overlays a KDE (Kernel Density Estimate) curve for a smoothed shape of the distribution.  

    📌 When to use a histogram:

To see how your data is distributed.

To check if the distribution is symmetric, skewed, etc.

To detect unusual values or gaps in data.

```python
# 3. Histogram - distribution of bill length
sns.histplot(
    penguins["bill_length_mm"], 
    bins=20, kde=True)    # bins are the bars and kde means kernel density estimate which shows the curve.
    plt.show()
```

     In Seaborn, a bar plot is used to show a summary statistic (like mean, median, sum) of a numeric variable for each category of a categorical variable.

     Unlike a histogram (which shows the distribution of a single numeric variable), a bar plot compares categories using aggregated values.

     What it does:

X-axis → Categorical variable (e.g., species, island).

Y-axis → Numeric variable’s aggregated value (default: mean).

Draws bars to represent the value.

     Can also use:

hue → to split bars into sub-categories (grouped bars).

estimator → to choose what statistic to show (mean, median, etc.).
   Example: estimator=np.median for median values.

ci → Size of the confidence interval (or None to hide it).

palette → Change colors.

     📌 When to use a bar plot:

Comparing averages or other summary stats between categories.

Showing differences between groups.

Comparing subgroups using hue.

```python
# 4. Bar plot (average body mass per species)
sns.barplot(
    x="species", 
    y="body_mass_g", 
    data=penguins)
plt.show()
```

    In Seaborn, a box plot (also called a whisker plot) is used to show the distribution of a dataset, highlighting its median, quartiles, and potential outliers.

    It’s great for comparing distributions across different categories.

    What a box plot shows:

A box plot summarizes data using five key numbers:

Minimum (lowest value excluding outliers)

-Q1 (First Quartile) — 25% of the data falls below this value

-Median — middle value (50% point)

-Q3 (Third Quartile) — 75% of the data falls below this value

-Maximum (highest value excluding outliers)

📌 Outliers are shown as points beyond the whiskers.

    How to read it:

Box = Middle 50% of the data (between Q1 and Q3)

Line inside box = Median

Whiskers = Range of non-outlier values

Dots outside whiskers = Outliers

    Key parameters:
x → Category (optional if you just want one box)

y → Numeric variable

hue → Add sub-groups with different colors

palette → Color scheme

orient → "h" for horizontal boxes

```python
# 5. Box plot
sns.boxplot(x="species", 
y="flipper_length_mm", 
data=penguins)
plt.show()
```

    In Seaborn, a count plot is used to show the number of occurrences of each category in a categorical variable.

    Think of it as:

📊 "A bar chart that automatically counts how many times each category appears in your data."

    Key points about Count Plot:

No need for a numeric y → Seaborn automatically calculates the counts.

hue → Allows you to split the bars by another category.

    Good for:

Seeing distribution of categories.

Comparing subgroup sizes.

Spotting imbalances in data.


```python
# Count plot - Number of penguins in each species
sns.countplot(
    x="species",         # Category to count
    data=penguins,       # Dataset
    palette="pastel"     # Color theme
)

plt.title("Number of Penguins per Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```
