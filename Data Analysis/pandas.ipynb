     Pandas-DataFrame And Series:
     
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
## Series
##A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
## Create a Series from dictionary
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
## Dataframe
## create a Dataframe from a dictionary oof list

data={
    'Name':['Krish','John','Jack'],
    'Age':[25,30,45],
    'City':['Bangalore','New York','Florida']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a Data frame From a List of Dictionaries

data=[
    {'Name':'Krish','Age':32,'City':'Bangalore'},
    {'Name':'John','Age':34,'City':'Bangalore'},
    {'Name':'Bappy','Age':32,'City':'Bangalore'},
    {'Name':'JAck','Age':32,'City':'Bangalore'}
    
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create a sample DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 40],
    "Salary": [50000, 60000, 75000, 90000]
}

# Load the data into a DataFrame
df = pd.DataFrame(data)

# Display the first few rows of the DataFrame
print(df.head())

# Filter rows where Age is greater than 30
df_filtered = df[df["Age"] > 30]

# Show the filtered DataFrame
print(df_filtered)
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Sort states by highest burglary rate
df.sort_values(by='BURGLARY(per_100k)', ascending=False)
```

```python
# Filter data for a specific year
df[df['YEAR'] == 2020]
```

```python
# States with IPC crime rate > 500
df[df['TOTAL IPC CRIMES(per_100k)'] > 500]
```

```python
# Top 5 states by murder rate
df.groupby('STATE/UT')['MURDER(per_100k)'] \
  .mean().sort_values(ascending=False).head(5)
```

```python
# Average theft vs burglary rates
df[['AUTO THEFT(per_100k)', 'BURGLARY(per_100k)']].mean()
```

```python
# First 10 unique states
df['STATE/UT'].unique()[:10]
```

```python
# Data for Delhi
df[df['STATE/UT'] == 'Delhi']
```

```python
# Data for year 2020
df[df['YEAR'] == 2020]
```

```python
# Sort by population (descending)
df.sort_values(by='POPULATION', ascending=False)
```

```python
# Max IPC crimes (per 100k) value
df['TOTAL IPC CRIMES(per_100k)'].max()
```

```python
# State with max IPC crimes in a single year
df.loc[df['TOTAL IPC CRIMES'].idxmax()]
```
