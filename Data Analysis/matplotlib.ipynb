     Data Visualization With Matplotlib:


Matplotlib is a powerful plotting library for Python that enables the creation of static, animated, and interactive visualizations. It is widely used for data visualization in data science and analytics. In this lesson, we will cover the basics of Matplotlib, including creating various types of plots and customizing them.

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[2,34,13,15,20]

##create a line plot
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y Axis')
plt.title("Simple Line Plot")
plt.show()
```

```python
##create a customized line plot

plt.plot(x,y,color='red',linestyle='--',marker='o',linewidth=3,markersize=9)
plt.grid(True)
```

```python
## Multiple Plots
## Sample data
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title("Plot 1")

plt.subplot(2,2,2)
plt.plot(y1,x,color='red')
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.plot(x,y2,color='blue')
plt.title("Plot 3")

plt.subplot(2,2,4)
plt.plot(x,y2,color='green')
plt.title("Plot 4")
```

```python
###Bar Plor
categories=['A','B','C','D','E']
values=[9,2,4,6,10]

##create a bar plot
plt.bar(categories,values,color='green')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Simple Bar Plot')
plt.show()
```

```python
# 1. Simple line plot
plt.plot([1, 2, 3, 4], [2, 4, 6, 8])
plt.show()

# 2. Add title and labels
plt.plot([1, 2, 3], [1, 4, 9])
plt.title("Simple Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
# 3. Change line style and color
plt.plot([1, 2, 3], [1, 4, 9], color='blue', linestyle='--', marker='o')
plt.show()
```

```python
# 4. Bar chart
plt.bar(['A', 'B', 'C'], [3, 7, 5], color='skyblue')
plt.show()
```

```python
# 5. Scatter plot
plt.scatter([5, 7, 8, 7], [5, 7, 8, 2], color='green')
plt.show()
```

```python
# 7. Multiple lines on same graph
x = [1, 2, 3]
plt.plot(x, [1, 2, 3], label="Line 1")
plt.plot(x, [1, 4, 9], label="Line 2")
plt.legend()
plt.show()
```
