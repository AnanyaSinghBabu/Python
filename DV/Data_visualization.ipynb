```python
# line plot
plt.plot(df['actual_price'])
plt.title("Price Trend")
plt.xlabel("Index")
plt.ylabel("Price")
plt.show()
```

```python
# scatter plot
plt.scatter(df['actual_price'], df['rating'])
plt.title("Price vs Rating")
plt.xlabel("actual_Price")
plt.ylabel("Rating")
plt.show()
```

```python
# bar plot
df['category'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Categories by Count")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# pie chart
df['category'].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%')
plt.title("Top 5 Categories Share")
plt.ylabel("")   # remove extra label
plt.show()
```

```python
#line plot
sns.lineplot(x=df.index, y=df['actual_price'])
plt.title("Price Trend")
plt.show()
```

```python
df['category'].value_counts().head(5).plot.pie(autopct='%1.1f%%', figsize=(6,6))
plt.title("Top 5 Categories Share")
plt.ylabel("")
plt.show()
```
