POSITIONAL ARGUMENT:

```python
#Positonal Argument
def ex1(a,b):
    print(a+b)
    
ex1(2,3)
```

```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

# Calling function with positional arguments
greet("Ananya", 24)
greet("Ravi", 25)
```

```python
def student_info(name, roll, course):
    # Printing inside the function
    print(f"Name: {name}, Roll: {roll}, Course: {course}")

# Call the function (outside the definition)
student_info("Payal", 101, "Data Analytics")
```

KEYWORD ARGUMENT:

```python
# Keyword argument
def ex2(name="ananya",age="24"):
    print(name,age)
    
ex2()
```

```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

# Call with keyword arguments
greet(name="Ananya", age=24)
greet(age=25, name="Ravi")  # order doesn’t matter here
```

```python
# Mixing positional + keyword argument
student_info("Ananya", roll=303, course="Data Analytics")

# rule: Positional arguments must always come before keyword arguments.
```

"Ananya" → positional argument (PA)
because it is passed without a keyword, directly based on position (goes to name).

roll=303 → keyword argument (KA)
because the parameter name (roll) is explicitly mentioned.

course="Machine Learning" → keyword argument (KA)
because the parameter name (course) is explicitly mentioned.

DEFAULT ARGUMENT:

```python
def ex3(a,b=4):
    print(a,b)

ex3(2)
```

```python
def greet(name, msg="Good Morning"):
    print("Hello", name, ",", msg)

# Without passing msg → uses default
greet("Ananya")

# Passing msg → overrides default
greet("Anjali", "How are you?")
```

```python
def employee(name, role="Intern", salary=20000):
    print(f"Name: {name}, Role: {role}, Salary: {salary}")

# Only pass name → role & salary take defaults
employee("Ananya")

# Override role
employee("Pratik", role="Data Analyst")

# Override both
employee("Anjali", role="Manager", salary=50000)
```

```python
# Default arguments must always come after positional ones in the function definition.

def func(a, b=10):   # ✅ correct
    pass

def func(a=10, b):   # ❌ error (default before non-default)
    pass
```

VARIABLE LENGTH ARGUMENT:

Variable-length arguments in Python allow a function to accept an arbitrary number of arguments. This is particularly useful when the exact number of arguments a function will receive is not known in advance. Python provides two main ways to handle variable-length arguments: 

    1.Positional Variable-Length Arguments (*args):

The *args syntax allows a function to accept any number of non-keyword (positional) arguments.(Collects values into a tuple.)

Inside the function, args is treated as a tuple containing all the positional arguments passed to the function.

This enables functions to be called with varying numbers of arguments, similar to how the built-in print() function works.

```python
def ex4(*a):
    print(a)
    
ex4(12,2,3,6,7,7)
```

```python
def add_numbers(*args):
    return sum(args)

print(add_numbers(10, 20))
print(add_numbers(5, 10, 15, 20))
 
```

```python
def show_numbers(*args):
    for num in args:
        print(num)

show_numbers(1, 2, 3)
show_numbers(10, 20, 30, 40, 50)
```

    2.Keyword Variable-Length Arguments (`: kwargs`)**:

The **kwargs syntax allows a function to accept any number of keyword arguments. 

Inside the function, kwargs is treated as a dictionary where keys are the argument names and values are the corresponding argument values.

This is useful for functions that might receive a flexible set of named parameters.

```python
def add_numbers(*nums):
    return sum(nums)

print(add_numbers(10, 20))          # 2 numbers
print(add_numbers(10, 20, 30, 40))  # 4 numbers
```

```python
def student_info(**details):
    for key, value in details.items():
        print(f"{key} = {value}")

student_info(name="Ananya", roll=101, course="Python")
student_info(name="Anjali", age=23, city="Delhi")
```

```python
# Mixing normal + *args + **kwargs

def details(college, *students, **info):
    print("College:", college)
    print("Students:", students)
    print("Other Info:", info)

details("IIT", "Ananya", "Anali", "Priya", location="Delhi", rank=5)
```
